```python
import spacy
from nltk import ngrams
nlp=spacy.load("en_core_web_sm")
doc=nlp("The quick brown fox jumps over the lazy dog")
bigrams=ngrams(doc,2)
print(list(bigrams))
```

```python
import spacy
from nltk import trigrams
nlp=spacy.load("en_core_web_sm")
doc=nlp("The quick brown fox jumps over the lazy dog")
trigrams=trigrams(doc)
print(list(trigrams))
```
